```python
def initialize_inventory(products):
    inventory = {}
    for product in products:
        quantity = int(input("Enter the quantity for " + product + ": "))
        inventory[product] = quantity
    return inventory
        
```

```python
products = ["banana", "kiwi", "orange", "apple"]
```

```python
initialize_inventory(products)
```

```python
inventory = initialize_inventory(products)
```

```python
def get_customer_orders():
    customer_orders = set()

    while True:
        product = input("Enter product name: ")

        if product == "done":
            break

        customer_orders.add(product)

    return customer_orders
```

```python
get_customer_orders()
```

```python
customer_orders = get_customer_orders()
```

```python
def update_inventory(customer_orders, inventory):
    for product in customer_orders:
        if product in inventory:
            inventory[product]-=1
        
    return inventory
```

```python
print(update_inventory(customer_orders, inventory))
```

```python
def calculate_order_statistics(customer_orders, products):
    total_products_ordered = len(customer_orders)
    percentage_unique_products = len(customer_orders)/len(products)*100
    return total_products_ordered, percentage_unique_products
```

```python
statistics = calculate_order_statistic(customer_orders, products)
print(statistics)
```

```python
def print_order_statistics(order_statistics):
    print("Total products ordered:", order_statistics[0])
    print("Percentage of unique products ordered:", order_statistics[1], "%")
```

```python
order_statistics = calculate_order_statistic(customer_orders, products)
```

```python
print_order_statistics(order_statistics)
```

```python
def print_updated_inventory(inventory):
    for product, quantity in inventory.items():
        print(product, ":", quantity)
```

```python
print_updated_inventory(inventory)
```

```python
products = ["banana", "kiwi", "orange", "apple"]

inventory = initialize_inventory(products)

customer_orders = get_customer_orders()

updated_inventory = update_inventory(customer_orders, inventory)

order_statistics = calculate_order_statistics(customer_orders, products)

print_order_statistics(order_statistics)

print_updated_inventory(updated_inventory)
```
